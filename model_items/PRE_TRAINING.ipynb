```python
import pandas as pd
import numpy as np

RR_FILE = "RR_clean.csv"
QOS_FILE = "QOS_clean.csv"

COMMON_FEATURES = [
    "reported_cqi",
    "buffer_log",
    "avg_rate_log",
    "estimated_rate_log",
    "pf_metric_log",
]

rr = pd.read_csv(RR_FILE)
qos = pd.read_csv(QOS_FILE)

print("=" * 90)
print("RR + QoS STATE FEATURE COMPARISON")
print("=" * 90)

print(f"RR rows  : {len(rr):,}")
print(f"QoS rows : {len(qos):,}")

# ============================================================
# COMMON FEATURES
# ============================================================

print("\nCommon features:")
print(COMMON_FEATURES)

print("\n" + "=" * 90)
print("RR SAMPLE")
print("=" * 90)

print(
    rr[COMMON_FEATURES + ["rr_priority"]]
    .head(10)
    .to_string(index=False)
)

print("\n" + "=" * 90)
print("QoS SAMPLE")
print("=" * 90)

print(
    qos[COMMON_FEATURES + ["qos_priority_scaled"]]
    .head(10)
    .to_string(index=False)
)

# ============================================================
# STATISTICAL COMPARISON
# ============================================================

rows = []

for feature in COMMON_FEATURES:
    rows.append({
        "feature": feature,

        "RR_min": rr[feature].min(),
        "RR_mean": rr[feature].mean(),
        "RR_median": rr[feature].median(),
        "RR_std": rr[feature].std(),
        "RR_max": rr[feature].max(),

        "QoS_min": qos[feature].min(),
        "QoS_mean": qos[feature].mean(),
        "QoS_median": qos[feature].median(),
        "QoS_std": qos[feature].std(),
        "QoS_max": qos[feature].max(),
    })

comparison = pd.DataFrame(rows)

print("\n" + "=" * 90)
print("COMMON FEATURE COMPARISON")
print("=" * 90)

print(comparison.to_string(index=False))

# ============================================================
# PRIORITY COMPARISON
# ============================================================

print("\n" + "=" * 90)
print("RR PRIORITY vs QoS PRIORITY")
print("=" * 90)

priority_comparison = pd.DataFrame({
    "RR_priority": rr["rr_priority"].describe(),
    "QoS_priority_scaled": qos["qos_priority_scaled"].describe()
})

print(priority_comparison.to_string())

print("\nRR priority values:")
print(sorted(rr["rr_priority"].dropna().unique()))

print("\nQoS scaled priority values:")
print(sorted(qos["qos_priority_scaled"].dropna().unique()))

# ============================================================
# NUMERICAL SAFETY
# ============================================================

print("\n" + "=" * 90)
print("NUMERICAL SAFETY")
print("=" * 90)

RR_FEATURES = COMMON_FEATURES + ["rr_priority"]
QOS_FEATURES = COMMON_FEATURES + ["qos_priority_scaled"]

for name, data, features in [
    ("RR", rr, RR_FEATURES),
    ("QoS", qos, QOS_FEATURES)
]:
    values = data[features].to_numpy(dtype=np.float64)

    print(
        f"{name:<5} "
        f"NaN={np.isnan(values).sum():,} "
        f"Inf={np.isinf(values).sum():,}"
    )

# ============================================================
# FINAL TRAINING STATE
# ============================================================

print("\n" + "=" * 90)
print("RR + QoS DQN STATE")
print("=" * 90)

print("""
Common features:
  reported_cqi
  buffer_log
  avg_rate_log
  estimated_rate_log
  pf_metric_log

Scheduler-specific priority:
  RR  -> rr_priority
  QoS -> qos_priority_scaled
""")

print("RR state features:")
print(COMMON_FEATURES + ["rr_priority"])

print("\nQoS state features:")
print(COMMON_FEATURES + ["qos_priority_scaled"])

print("\n" + "=" * 90)
print(qos.info())
print("\n" + "=" * 90)
print(qos.info())
```
